## Student Attendance Analytics with Pandas

```python
import pandas as pd
import numpy as np
```

#### Section 0: Data Loading

```python
df = pd.read_excel("aids_attendance.xlsx")
display(df)
```

#### Section 1: Data Inspection

```python
print("First 10 rows:")
display(df.head(10))
print("\nShape:", df.shape)
print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values in each column:")
print(df.isna().sum())
print("\nRows where Last Present Date is Never:", df["Last Present Date"].astype(str).str.strip().str.lower().eq("never").sum())
```

```python
df = df.rename(columns={"Roll No":"roll_no", "Student Name":"student_name", "Attendance %":"attendance_pct", "Total Classes":"total_classes", "Present":"present", "Absent":"absent", "Excused":"excused", "Status":"status", "Classes Needed (75%)":"classes_needed_75", "Last Present Date":"last_present_date", "Consecutive Absences":"consecutive_absences"})
print(df.columns.tolist())
```

```python
df["attendance_pct"] = pd.to_numeric(df["attendance_pct"].astype(str).str.replace("%", "", regex=False).str.strip(), errors="coerce")
df["last_present_date"] = pd.to_datetime(df["last_present_date"].replace("Never", pd.NA), errors="coerce")
print(df[["attendance_pct", "last_present_date"]].head(10))
print("\nAttendance percentage is stored on a 0–100 scale.")
```

#### Section 2: Filtering & Selection

```python
critical = df.loc[df["status"].eq("Critical")].sort_values("attendance_pct", ascending=True)
display(critical)
```

```python
fully_absent = df.loc[df["present"].eq(0), ["roll_no", "student_name", "consecutive_absences"]]
display(fully_absent)
```

```python
print("(a) Safe students using .loc:")
display(df.loc[df["status"].eq("Safe")])
print("(b) First 15 rows and columns 2–6 using .iloc:")
display(df.iloc[:15, 1:6])
```

```python
display(df.loc[df["consecutive_absences"].ge(5), ["student_name", "status", "last_present_date"]])
```

### Section 3: Grouping & Aggregation 

```python
status_summary = df.groupby("status").agg(student_count=("roll_no", "count"), average_attendance_pct=("attendance_pct", "mean"), max_consecutive_absences=("consecutive_absences", "max"))
display(status_summary)
```

```python
df["attendance_category"] = pd.cut(df["attendance_pct"], bins=[-0.01, 59.999, 74.999, 89.999, 100], labels=["Poor", "Average", "Good", "Excellent"], include_lowest=True)
print(df.groupby("attendance_category", observed=False).size().rename("student_count"))
```

```python
totals = pd.Series({"total_classes_conducted": df["total_classes"].sum(), "total_present": df["present"].sum(), "total_absent": df["absent"].sum()})
display(totals)
```

```python
display(df.groupby("status").agg(present_min=("present", "min"), present_max=("present", "max"), present_mean=("present", "mean"), absent_sum=("absent", "sum")))
```

### Section 4: Feature Engineering & Transformation

```python
df["risk_score"] = (df["absent"] * 0.4) + (df["consecutive_absences"] * 0.6)
display(df.nlargest(10, "risk_score")[["roll_no", "student_name", "absent", "consecutive_absences", "risk_score"]])
```

```python
df["classes_needed_numeric"] = pd.to_numeric(df["classes_needed_75"].astype(str).str.extract(r"(\d+)")[0], errors="coerce").fillna(0).astype(int)
display(df[["classes_needed_75", "classes_needed_numeric"]].head(15))
```

```python
df["attendance_flag"] = df["attendance_pct"].apply(lambda x: "Defaulter" if x < 75 else "Safe")
display(df[["student_name", "attendance_pct", "attendance_flag"]])
```

```python
df["attendance_range"] = pd.cut(df["attendance_pct"], bins=[0, 50, 75, 90, 100], labels=["Critical", "Warning", "Good", "Excellent"], include_lowest=True, right=True)
display(df[["student_name", "attendance_pct", "attendance_range"]])
```

### Section 5: Advanced Analysis 

```python
pivot = pd.pivot_table(df, index="status", columns="attendance_category", values="roll_no", aggfunc="count", fill_value=0, observed=False)
display(pivot)
```

```python
df["base_name"] = df["student_name"].astype(str).str.strip().str.upper().str.split().str[0].str.replace(r"[^A-Z]", "", regex=True)
duplicate_bases = df["base_name"].duplicated(keep=False)
display(df.loc[duplicate_bases].sort_values("base_name", na_position="last"))
```

```python
# Hypothetical details for demonstrating an inner join; replace these sample values if your instructor specifies others.
student_details = pd.DataFrame({
    "Roll No": df["roll_no"].drop_duplicates(),
    "Branch": "AI&DS",
    "Semester": "Not specified"
})
merged = df.merge(student_details, left_on="roll_no", right_on="Roll No", how="inner")
display(merged)
```

```python
df.to_csv("attendance_analytics_completed.csv", index=False)
df.loc[df["status"].eq("Critical")].to_csv("critical_students.csv", index=False)
status_summary.to_csv("status_summary.csv")
df.nlargest(10, "risk_score").to_csv("top_10_risk_students.csv", index=False)
print("Exported: attendance_analytics_completed.csv, critical_students.csv, status_summary.csv, top_10_risk_students.csv")
```
